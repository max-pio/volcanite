```python
import pyvista as pv
import numpy as np
from timeit import default_timer as timer
import PIL.Image as Image
import matplotlib.pyplot as plt
```

```python
def read_from_NRRD(path):
    with open(path, "rb") as file:
        # read header
        shape_str = file.readline()[:-1].decode('ascii').split() # dim_x dim_y dim_z
        type = file.readline()[:-1].decode('ascii')              # dtype
        # read binary
        volume = np.fromfile(file, dtype=type)
        volume = volume.reshape([int(shape_str[0]), int(shape_str[1]), int(shape_str[2])])
    return volume
```

```python
volume = read_from_NRRD("data/cells_frame000.raw");
print("Read volume with dimension " + str(volume.shape) + " and data type " + str(volume.dtype))
```

```python
fig = plt.figure()
plt.imshow(volume[:][:][128], cmap='gray', interpolation='nearest')
plt.show()
```
